```python
run1_raw = """
0.0875 0.7244
0.1 0.7129
0.1125 0.7083
0.125 0.7015
0.1375 0.6986
0.15 0.6947
0.1625 0.691
0.175 0.6903
0.1875 0.6908
0.2 0.6958
0.2125 0.6938
0.225 0.6878
0.2375 0.6859
0.25 0.6871
0.2625 0.6899
0.275 0.6882
0.2875 0.6881
0.3 0.687
0.3125 0.6894
0.325 0.6882
0.3375 0.6877
0.35 0.684
0.3625 0.6862
0.375 0.6816
0.3875 0.682
0.4 0.6775
0.425 0.6833
0.45 0.6838
0.475 0.685
0.5 0.6853
0.525 0.686
0.55 0.6867
0.575 0.6846
0.6 0.6821
0.625 0.6789
0.65 0.6774
0.675 0.6795
0.7 0.6809
0.725 0.682
0.75 0.6813
0.775 0.6806
0.8 0.6803
0.85 0.6795
0.9 0.6783
0.95 0.6774
1 0.6755
1.05 0.6767
1.1 0.677
1.15 0.679
1.2 0.6769
1.25 0.6785
1.3 0.678
1.35 0.6794
1.4 0.6784
1.45 0.6777
1.5 0.6763
1.55 0.6747
1.6 0.6735
1.7 0.6743
1.8 0.6743
1.9 0.674
2 0.6729
2.1 0.6725
2.2 0.6721
2.3 0.6727
2.4 0.6714
2.5 0.6708
2.6 0.6695
2.7 0.6701
2.8 0.6701
2.9 0.67
3 0.6706
3.1 0.6712
3.2 0.6715
3.4 0.6698
3.6 0.6686
3.8 0.6678
4 0.6664
4.2 0.6652
4.4 0.6647
4.6 0.6643
4.8 0.6645
5 0.6636
5.2 0.6636
5.4 0.6625
5.6 0.6616
5.8 0.6607
6 0.6606
6.2 0.6597
6.4 0.6586
6.8 0.6564
7.2 0.6559
7.6 0.6549
8 0.6543
8.4 0.6526
8.8 0.6513
9.2 0.6497
9.6 0.6487
10 0.6475
10.4 0.6459
10.8 0.6444
11.2 0.6435
11.6 0.6422
12 0.641
12.4 0.64
12.8 0.6394
13.6 0.6377
14.4 0.6355
15.2 0.633
16 0.6313
16.8 0.6291
17.6 0.6272
18.4 0.625
19.2 0.6233
20 0.6216
20.8 0.6197
21.6 0.6181
22.4 0.6161
23.2 0.6148
24 0.6128
24.8 0.6112
25.6 0.6092
27.2 0.6058
28.8 0.602
30.4 0.5989
32 0.5955
33.6 0.5925
35.2 0.589
36.8 0.5858
38.4 0.5826
40 0.5798
41.6 0.5772
43.2 0.5746
44.8 0.5717
46.4 0.5688
48 0.5658
49.6 0.5628
51.2 0.5601
54.4 0.5547
57.6 0.5497
60.8 0.5445
64 0.5394
67.2 0.5345
70.4 0.5296
73.6 0.5249
76.8 0.5201
80 0.5155
83.2 0.5108
86.4 0.5062
89.6 0.5018
92.8 0.4975
96 0.4934
99.2 0.4892
102.4 0.4849
108.8 0.4763
115.2 0.468
121.6 0.4601
128 0.4523
134.4 0.4447
140.8 0.437
147.2 0.4299
153.6 0.4226
160 0.4157
166.4 0.4088
172.8 0.4021
179.2 0.3956
185.6 0.3893
192 0.3831
198.4 0.377
204.8 0.371
217.6 0.3595
230.4 0.3484
243.2 0.3379
256 0.3277
268.8 0.3181
281.6 0.3089
294.4 0.3001
307.2 0.2917
320 0.2835
332.8 0.2758
345.6 0.2683
358.4 0.2611
371.2 0.2542
384 0.2476
396.8 0.2412
409.6 0.2351
435.2 0.2236
460.8 0.2131
486.4 0.2035
512 0.1946
537.6 0.1863
563.2 0.1789
588.8 0.1723
614.4 0.1662
640 0.1604
665.6 0.1551
691.2 0.1502
716.8 0.1457
742.4 0.1417
768 0.1381
793.6 0.1348
819.2 0.1318
870.4 0.1263
921.6 0.1215
972.8 0.1172
1024 0.1134
1075 0.11
1126 0.1071
1178 0.1046
1229 0.1024
1280 0.1006
1331 0.099
1382 0.09748
1434 0.09599
1485 0.09466
1536 0.09366
1587 0.0931
1638 0.09276
1741 0.09206
1843 0.09154
1946 0.09139
2048 0.09146
2150 0.09104
2253 0.09013
2355 0.08915
2458 0.08853
2560 0.08788
2662 0.08687
2765 0.08534
2867 0.08357
2970 0.08169
3072 0.07975
3174 0.07832
3277 0.07697
3482 0.07436
3686 0.07147
3891 0.06852
4096 0.06604
4301 0.06418
4506 0.06269
4710 0.06076
4915 0.05814
5120 0.05627
5325 0.0559
5530 0.05649
5734 0.05678
5939 0.05648
6144 0.05526
6349 0.05357
6554 0.05189
6963 0.04917
7373 0.04728
7782 0.04553
8192 0.04393
8602 0.04213
9011 0.04033
9421 0.03917
9830 0.03764
1.02E+04 0.03559
1.07E+04 0.0336
1.11E+04 0.03314
1.15E+04 0.03349
1.19E+04 0.03426
1.23E+04 0.03393
1.27E+04 0.03317
1.31E+04 0.03153
1.39E+04 0.02844
1.48E+04 0.02612
1.56E+04 0.02472
1.64E+04 0.02375
1.72E+04 0.02293
1.80E+04 0.02227
1.88E+04 0.0212
1.97E+04 0.02043
2.05E+04 0.01984
2.13E+04 0.01892
2.21E+04 0.01797
2.29E+04 0.01651
2.38E+04 0.01596
2.46E+04 0.01601
2.54E+04 0.01585
2.62E+04 0.01532
2.79E+04 0.0123
2.95E+04 0.011
3.11E+04 0.01043
3.28E+04 0.01028
3.44E+04 0.00955
3.60E+04 0.00869
3.77E+04 0.008702
3.93E+04 0.008941
4.10E+04 0.009474
4.26E+04 0.009007
4.42E+04 0.008202
4.59E+04 0.007521
4.75E+04 0.00707
4.92E+04 0.006757
5.08E+04 0.006155
5.24E+04 0.005876
5.57E+04 0.006111
5.90E+04 0.006727
6.23E+04 0.007299
6.55E+04 0.007385
6.88E+04 0.007468
7.21E+04 0.007089
7.54E+04 0.006774
7.86E+04 0.006946
8.19E+04 0.008077
8.52E+04 0.00872
8.85E+04 0.008427
9.18E+04 0.007553
9.50E+04 0.006931
9.83E+04 0.006819
1.02E+05 0.007372
1.05E+05 0.008208
1.11E+05 0.009441
1.18E+05 0.01063
1.25E+05 0.01217
1.31E+05 0.01349
1.38E+05 0.01466
1.44E+05 0.01493
1.51E+05 0.01471
1.57E+05 0.01322
1.64E+05 0.01158
1.70E+05 0.009811
1.77E+05 0.009157
1.84E+05 0.008865
1.90E+05 0.009338
1.97E+05 0.008989
2.03E+05 0.009205
2.10E+05 0.00887
2.23E+05 0.008684
2.36E+05 0.008565
2.49E+05 0.009333
2.62E+05 0.01053
2.75E+05 0.01151
2.88E+05 0.01159
3.02E+05 0.01149
3.15E+05 0.01104
3.28E+05 0.01107
3.41E+05 0.01083
3.54E+05 0.0103
3.67E+05 0.009866
3.80E+05 0.009562
3.93E+05 0.009924
4.06E+05 0.01025
4.19E+05 0.01157
4.46E+05 0.01259
4.72E+05 0.01291
4.98E+05 0.01239
5.24E+05 0.01261
5.51E+05 0.01257
5.77E+05 0.01197
6.03E+05 0.01109
6.29E+05 0.01171
6.55E+05 0.01319
6.82E+05 0.01419
7.08E+05 0.0134
7.34E+05 0.01215
7.60E+05 0.01198
7.86E+05 0.01307
8.13E+05 0.01368
8.39E+05 0.01384
8.91E+05 0.01267
9.44E+05 0.0114
9.96E+05 0.009687
1.05E+06 0.008373
1.10E+06 0.007344
1.15E+06 0.007549
1.21E+06 0.007133
1.26E+06 0.006555
1.31E+06 0.005412
1.36E+06 0.004983
1.42E+06 0.004762
1.47E+06 0.003914
1.52E+06 0.00296
1.57E+06 0.00222
1.63E+06 0.001477
1.68E+06 0.001021
1.78E+06 0
1.89E+06 0
1.99E+06 0
2.10E+06 0
2.20E+06 0
2.31E+06 0
2.41E+06 0
2.52E+06 0
2.62E+06 0
2.73E+06 0
2.83E+06 0
2.94E+06 0
3.04E+06 0
3.15E+06 0
3.25E+06 0
3.36E+06 0
3.57E+06 0
3.78E+06 0
3.99E+06 0
4.19E+06 0
4.40E+06 0
4.61E+06 0
4.82E+06 0
5.03E+06 0
5.24E+06 0
5.45E+06 0
5.66E+06 0
5.87E+06 0
6.08E+06 0
6.29E+06 0
6.50E+06 0
6.71E+06 0
7.13E+06 0
7.55E+06 0
7.97E+06 0
8.39E+06 0
8.81E+06 0
9.23E+06 0
9.65E+06 0
1.01E+07 0
1.05E+07 0
1.09E+07 0
1.13E+07 0
1.17E+07 0
1.22E+07 0
1.26E+07 0
1.30E+07 0
1.34E+07 0
1.43E+07 0
1.51E+07 0
1.59E+07 0
1.68E+07 0
1.76E+07 0
1.85E+07 0
1.93E+07 0
2.01E+07 0
2.10E+07 0
2.18E+07 0
2.27E+07 0
2.35E+07 0
2.43E+07 0
2.52E+07 0
2.60E+07 0
2.68E+07 0
2.85E+07 0
3.02E+07 0
3.19E+07 0
3.36E+07 0
3.52E+07 0
3.69E+07 0
3.86E+07 0
4.03E+07 0
4.19E+07 0
4.36E+07 0
4.53E+07 0
4.70E+07 0
4.87E+07 0
5.03E+07 0
5.20E+07 0
5.37E+07 0
5.70E+07 0
6.04E+07 0
6.38E+07 0
6.71E+07 0
7.05E+07 0
7.38E+07 0
7.72E+07 0
8.05E+07 0
8.39E+07 0
8.72E+07 0
9.06E+07 0
9.40E+07 0
9.73E+07 0
1.01E+08 0
1.04E+08 0
1.07E+08 0
1.14E+08 0
1.21E+08 0
1.28E+08 0
1.34E+08 0
1.41E+08 0
1.48E+08 0
1.54E+08 0
1.61E+08 0
1.68E+08 0
1.75E+08 0
1.81E+08 0
1.88E+08 0
1.95E+08 0
2.01E+08 0
2.08E+08 0
2.15E+08 0
"""

run2_raw = """
0.0875 0.7395
0.1 0.7314
0.1125 0.7249
0.125 0.7187
0.1375 0.7165
0.15 0.7139
0.1625 0.7124
0.175 0.7122
0.1875 0.7111
0.2 0.7105
0.2125 0.7084
0.225 0.7086
0.2375 0.7053
0.25 0.7052
0.2625 0.7036
0.275 0.7044
0.2875 0.7057
0.3 0.7063
0.3125 0.704
0.325 0.7002
0.3375 0.7006
0.35 0.7039
0.3625 0.7055
0.375 0.7045
0.3875 0.704
0.4 0.7035
0.425 0.7018
0.45 0.6996
0.475 0.6993
0.5 0.698
0.525 0.6978
0.55 0.6978
0.575 0.6975
0.6 0.7
0.625 0.6991
0.65 0.6997
0.675 0.6973
0.7 0.6986
0.725 0.6977
0.75 0.6987
0.775 0.6984
0.8 0.6983
0.85 0.6965
0.9 0.694
0.95 0.696
1 0.694
1.05 0.6953
1.1 0.6948
1.15 0.6954
1.2 0.6941
1.25 0.6929
1.3 0.693
1.35 0.6949
1.4 0.6953
1.45 0.6945
1.5 0.6924
1.55 0.6915
1.6 0.6917
1.7 0.691
1.8 0.69
1.9 0.6903
2 0.6906
2.1 0.6912
2.2 0.6908
2.3 0.691
2.4 0.6901
2.5 0.6898
2.6 0.6891
2.7 0.6888
2.8 0.6869
2.9 0.6867
3 0.6869
3.1 0.6873
3.2 0.6874
3.4 0.6859
3.6 0.6856
3.8 0.6845
4 0.6836
4.2 0.6826
4.4 0.6822
4.6 0.6814
4.8 0.6814
5 0.6809
5.2 0.6805
5.4 0.6795
5.6 0.6789
5.8 0.678
6 0.6774
6.2 0.6763
6.4 0.6764
6.8 0.6752
7.2 0.6739
7.6 0.6725
8 0.6715
8.4 0.671
8.8 0.6699
9.2 0.6685
9.6 0.6673
10 0.6662
10.4 0.6654
10.8 0.6642
11.2 0.6631
11.6 0.6623
12 0.661
12.4 0.6598
12.8 0.6586
13.6 0.6569
14.4 0.6549
15.2 0.6526
16 0.6509
16.8 0.6493
17.6 0.6473
18.4 0.6457
19.2 0.6437
20 0.642
20.8 0.64
21.6 0.6382
22.4 0.6367
23.2 0.6348
24 0.6333
24.8 0.6317
25.6 0.6302
27.2 0.6272
28.8 0.6241
30.4 0.6212
32 0.6183
33.6 0.6156
35.2 0.6128
36.8 0.6098
38.4 0.6068
40 0.6039
41.6 0.6012
43.2 0.5988
44.8 0.5966
46.4 0.5941
48 0.5917
49.6 0.589
51.2 0.5867
54.4 0.5817
57.6 0.5768
60.8 0.5722
64 0.5678
67.2 0.5635
70.4 0.559
73.6 0.5547
76.8 0.5504
80 0.5464
83.2 0.5423
86.4 0.5384
89.6 0.5344
92.8 0.5305
96 0.5265
99.2 0.5227
102.4 0.519
108.8 0.5118
115.2 0.5045
121.6 0.4975
128 0.4906
134.4 0.4839
140.8 0.4772
147.2 0.4707
153.6 0.4643
160 0.458
166.4 0.4518
172.8 0.4458
179.2 0.4399
185.6 0.4341
192 0.4285
198.4 0.423
204.8 0.4177
217.6 0.4072
230.4 0.3969
243.2 0.3869
256 0.3774
268.8 0.3683
281.6 0.3596
294.4 0.3512
307.2 0.343
320 0.3351
332.8 0.3275
345.6 0.3204
358.4 0.3133
371.2 0.3066
384 0.3001
396.8 0.2941
409.6 0.2883
435.2 0.2774
460.8 0.2671
486.4 0.2575
512 0.2486
537.6 0.2403
563.2 0.2327
588.8 0.2256
614.4 0.2191
640 0.213
665.6 0.2074
691.2 0.202
716.8 0.197
742.4 0.1923
768 0.1879
793.6 0.1838
819.2 0.1801
870.4 0.1734
921.6 0.1675
972.8 0.1623
1024 0.1578
1075 0.1538
1126 0.1503
1178 0.1472
1229 0.1444
1280 0.1419
1331 0.1395
1382 0.1375
1434 0.1355
1485 0.1336
1536 0.1318
1587 0.1303
1638 0.129
1741 0.1267
1843 0.1249
1946 0.1231
2048 0.1216
2150 0.1202
2253 0.119
2355 0.1176
2458 0.1158
2560 0.1139
2662 0.1124
2765 0.1109
2867 0.1094
2970 0.1076
3072 0.1059
3174 0.1045
3277 0.1034
3482 0.1016
3686 0.09942
3891 0.09694
4096 0.09398
4301 0.09097
4506 0.08804
4710 0.08522
4915 0.08271
5120 0.08046
5325 0.07862
5530 0.07699
5734 0.07524
5939 0.07335
6144 0.07149
6349 0.07
6554 0.06823
6963 0.06451
7373 0.06085
7782 0.05808
8192 0.05574
8602 0.05337
9011 0.05014
9421 0.04613
9830 0.0431
1.02E+04 0.04165
1.07E+04 0.04212
1.11E+04 0.04197
1.15E+04 0.04053
1.19E+04 0.0388
1.23E+04 0.03757
1.27E+04 0.03713
1.31E+04 0.03599
1.39E+04 0.03302
1.48E+04 0.02977
1.56E+04 0.02679
1.64E+04 0.02387
1.72E+04 0.0226
1.80E+04 0.02232
1.88E+04 0.02274
1.97E+04 0.02168
2.05E+04 0.02077
2.13E+04 0.02061
2.21E+04 0.02121
2.29E+04 0.02141
2.38E+04 0.02065
2.46E+04 0.0191
2.54E+04 0.01728
2.62E+04 0.01522
2.79E+04 0.01304
2.95E+04 0.0111
3.11E+04 0.0105
3.28E+04 0.01036
3.44E+04 0.01076
3.60E+04 0.01001
3.77E+04 0.008971
3.93E+04 0.007864
4.10E+04 0.007747
4.26E+04 0.00777
4.42E+04 0.007691
4.59E+04 0.007478
4.75E+04 0.006859
4.92E+04 0.006404
5.08E+04 0.006044
5.24E+04 0.005911
5.57E+04 0.006134
5.90E+04 0.005893
6.23E+04 0.005543
6.55E+04 0.005353
6.88E+04 0.00624
7.21E+04 0.006769
7.54E+04 0.007122
7.86E+04 0.006372
8.19E+04 0.006159
8.52E+04 0.006043
8.85E+04 0.006198
9.18E+04 0.006292
9.50E+04 0.006059
9.83E+04 0.006081
1.02E+05 0.005825
1.05E+05 0.006216
1.11E+05 0.006978
1.18E+05 0.008448
1.25E+05 0.009539
1.31E+05 0.01101
1.38E+05 0.01122
1.44E+05 0.01065
1.51E+05 0.009575
1.57E+05 0.009087
1.64E+05 0.008486
1.70E+05 0.007386
1.77E+05 0.006544
1.84E+05 0.006996
1.90E+05 0.00776
1.97E+05 0.008406
2.03E+05 0.007655
2.10E+05 0.007081
2.23E+05 0.00616
2.36E+05 0.006197
2.49E+05 0.006809
2.62E+05 0.006969
2.75E+05 0.007277
2.88E+05 0.007178
3.02E+05 0.00733
3.15E+05 0.008028
3.28E+05 0.008973
3.41E+05 0.01022
3.54E+05 0.009706
3.67E+05 0.008857
3.80E+05 0.008214
3.93E+05 0.008971
4.06E+05 0.009795
4.19E+05 0.01062
4.46E+05 0.01126
4.72E+05 0.01129
4.98E+05 0.01031
5.24E+05 0.009789
5.51E+05 0.009812
5.77E+05 0.0108
6.03E+05 0.01119
6.29E+05 0.01107
6.55E+05 0.01066
6.82E+05 0.01082
7.08E+05 0.01174
7.34E+05 0.01229
7.60E+05 0.01246
7.86E+05 0.01192
8.13E+05 0.01193
8.39E+05 0.01185
8.91E+05 0.0126
9.44E+05 0.01288
9.96E+05 0.01303
1.05E+06 0.01217
1.10E+06 0.01128
1.15E+06 0.01094
1.21E+06 0.01042
1.26E+06 0.01009
1.31E+06 0.00875
1.36E+06 0.008271
1.42E+06 0.007684
1.47E+06 0.00766
1.52E+06 0.007186
1.57E+06 0.005824
1.63E+06 0.003395
1.68E+06 0.001696
1.78E+06 0
1.89E+06 0
1.99E+06 0
2.10E+06 0
2.20E+06 0
2.31E+06 0
2.41E+06 0
2.52E+06 0
2.62E+06 0
2.73E+06 0
2.83E+06 0
2.94E+06 0
3.04E+06 0
3.15E+06 0
3.25E+06 0
3.36E+06 0
3.57E+06 0
3.78E+06 0
3.99E+06 0
4.19E+06 0
4.40E+06 0
4.61E+06 0
4.82E+06 0
5.03E+06 0
5.24E+06 0
5.45E+06 0
5.66E+06 0
5.87E+06 0
6.08E+06 0
6.29E+06 0
6.50E+06 0
6.71E+06 0
7.13E+06 0
7.55E+06 0
7.97E+06 0
8.39E+06 0
8.81E+06 0
9.23E+06 0
9.65E+06 0
1.01E+07 0
1.05E+07 0
1.09E+07 0
1.13E+07 0
1.17E+07 0
1.22E+07 0
1.26E+07 0
1.30E+07 0
1.34E+07 0
1.43E+07 0
1.51E+07 0
1.59E+07 0
1.68E+07 0
1.76E+07 0
1.85E+07 0
1.93E+07 0
2.01E+07 0
2.10E+07 0
2.18E+07 0
2.27E+07 0
2.35E+07 0
2.43E+07 0
2.52E+07 0
2.60E+07 0
2.68E+07 0
2.85E+07 0
3.02E+07 0
3.19E+07 0
3.36E+07 0
3.52E+07 0
3.69E+07 0
3.86E+07 0
4.03E+07 0
4.19E+07 0
4.36E+07 0
4.53E+07 0
4.70E+07 0
4.87E+07 0
5.03E+07 0
5.20E+07 0
5.37E+07 0
5.70E+07 0
6.04E+07 0
6.38E+07 0
6.71E+07 0
7.05E+07 0
7.38E+07 0
7.72E+07 0
8.05E+07 0
8.39E+07 0
8.72E+07 0
9.06E+07 0
9.40E+07 0
9.73E+07 0
1.01E+08 0
1.04E+08 0
1.07E+08 0
1.14E+08 0
1.21E+08 0
1.28E+08 0
1.34E+08 0
1.41E+08 0
1.48E+08 0
1.54E+08 0
1.61E+08 0
1.68E+08 0
1.75E+08 0
1.81E+08 0
1.88E+08 0
1.95E+08 0
2.01E+08 0
2.08E+08 0
2.15E+08 0
"""

run3_raw = """
0.0875 0.7458
0.1 0.7453
0.1125 0.7388
0.125 0.7331
0.1375 0.7233
0.15 0.7222
0.1625 0.72
0.175 0.7237
0.1875 0.7215
0.2 0.7182
0.2125 0.7167
0.225 0.7179
0.2375 0.7209
0.25 0.7208
0.2625 0.7183
0.275 0.716
0.2875 0.7106
0.3 0.7115
0.3125 0.7095
0.325 0.713
0.3375 0.7099
0.35 0.7112
0.3625 0.711
0.375 0.7099
0.3875 0.7073
0.4 0.7046
0.425 0.7051
0.45 0.706
0.475 0.7065
0.5 0.7062
0.525 0.7044
0.55 0.7036
0.575 0.7057
0.6 0.7057
0.625 0.7051
0.65 0.7061
0.675 0.7062
0.7 0.7063
0.725 0.7044
0.75 0.7025
0.775 0.7017
0.8 0.6999
0.85 0.7021
0.9 0.7022
0.95 0.7026
1 0.7019
1.05 0.7008
1.1 0.6996
1.15 0.7004
1.2 0.7005
1.25 0.7003
1.3 0.6988
1.35 0.7007
1.4 0.7015
1.45 0.7023
1.5 0.6998
1.55 0.6989
1.6 0.6978
1.7 0.6979
1.8 0.6971
1.9 0.6972
2 0.6969
2.1 0.6969
2.2 0.696
2.3 0.6968
2.4 0.6961
2.5 0.6958
2.6 0.6945
2.7 0.695
2.8 0.6951
2.9 0.6944
3 0.6932
3.1 0.6921
3.2 0.6927
3.4 0.6931
3.6 0.6923
3.8 0.6909
4 0.69
4.2 0.6896
4.4 0.6885
4.6 0.6878
4.8 0.687
5 0.6864
5.2 0.6854
5.4 0.6844
5.6 0.6833
5.8 0.6831
6 0.6828
6.2 0.683
6.4 0.6816
6.8 0.68
7.2 0.6782
7.6 0.6775
8 0.6764
8.4 0.6749
8.8 0.6729
9.2 0.6716
9.6 0.6701
10 0.6695
10.4 0.668
10.8 0.6668
11.2 0.6649
11.6 0.663
12 0.6615
12.4 0.6603
12.8 0.6596
13.6 0.6579
14.4 0.6562
15.2 0.6543
16 0.6521
16.8 0.6498
17.6 0.6477
18.4 0.6458
19.2 0.6443
20 0.6423
20.8 0.6402
21.6 0.6378
22.4 0.6358
23.2 0.6337
24 0.6323
24.8 0.6298
25.6 0.628
27.2 0.624
28.8 0.6207
30.4 0.6172
32 0.6136
33.6 0.6104
35.2 0.607
36.8 0.6036
38.4 0.6004
40 0.5974
41.6 0.5945
43.2 0.5913
44.8 0.5879
46.4 0.5844
48 0.5815
49.6 0.5784
51.2 0.5758
54.4 0.5697
57.6 0.5644
60.8 0.559
64 0.5538
67.2 0.5486
70.4 0.5432
73.6 0.5379
76.8 0.5326
80 0.5275
83.2 0.5226
86.4 0.5179
89.6 0.5129
92.8 0.5083
96 0.5035
99.2 0.4988
102.4 0.4941
108.8 0.4851
115.2 0.4763
121.6 0.4679
128 0.4597
134.4 0.4517
140.8 0.444
147.2 0.4363
153.6 0.4288
160 0.4214
166.4 0.4143
172.8 0.4073
179.2 0.4005
185.6 0.3938
192 0.3875
198.4 0.3811
204.8 0.3751
217.6 0.3633
230.4 0.3521
243.2 0.3413
256 0.331
268.8 0.3211
281.6 0.3117
294.4 0.303
307.2 0.2947
320 0.2869
332.8 0.2793
345.6 0.272
358.4 0.2652
371.2 0.2589
384 0.2528
396.8 0.2468
409.6 0.2412
435.2 0.2307
460.8 0.2212
486.4 0.2125
512 0.2045
537.6 0.1971
563.2 0.1904
588.8 0.1844
614.4 0.1789
640 0.1735
665.6 0.1685
691.2 0.1638
716.8 0.1596
742.4 0.1558
768 0.1522
793.6 0.1489
819.2 0.146
870.4 0.1413
921.6 0.1373
972.8 0.134
1024 0.1313
1075 0.1288
1126 0.1265
1178 0.1242
1229 0.1221
1280 0.1202
1331 0.1187
1382 0.1174
1434 0.1165
1485 0.1156
1536 0.1148
1587 0.1137
1638 0.1127
1741 0.111
1843 0.1097
1946 0.1081
2048 0.1063
2150 0.1046
2253 0.1029
2355 0.101
2458 0.09853
2560 0.0957
2662 0.09307
2765 0.09092
2867 0.0896
2970 0.0889
3072 0.08849
3174 0.08759
3277 0.08617
3482 0.08241
3686 0.07879
3891 0.07536
4096 0.0725
4301 0.07032
4506 0.06815
4710 0.06561
4915 0.06297
5120 0.06085
5325 0.05937
5530 0.05768
5734 0.0561
5939 0.05391
6144 0.05191
6349 0.04951
6554 0.04751
6963 0.04383
7373 0.04078
7782 0.03822
8192 0.03585
8602 0.03391
9011 0.03205
9421 0.03001
9830 0.02869
1.02E+04 0.0278
1.07E+04 0.02654
1.11E+04 0.02429
1.15E+04 0.02275
1.19E+04 0.02285
1.23E+04 0.02347
1.27E+04 0.02264
1.31E+04 0.0211
1.39E+04 0.01847
1.48E+04 0.01652
1.56E+04 0.01531
1.64E+04 0.01434
1.72E+04 0.0142
1.80E+04 0.01386
1.88E+04 0.01298
1.97E+04 0.01181
2.05E+04 0.01086
2.13E+04 0.009234
2.21E+04 0.008168
2.29E+04 0.007167
2.38E+04 0.00774
2.46E+04 0.008043
2.54E+04 0.008467
2.62E+04 0.008223
2.79E+04 0.00773
2.95E+04 0.007203
3.11E+04 0.007305
3.28E+04 0.007162
3.44E+04 0.007004
3.60E+04 0.006201
3.77E+04 0.005768
3.93E+04 0.005395
4.10E+04 0.005497
4.26E+04 0.005451
4.42E+04 0.005262
4.59E+04 0.004824
4.75E+04 0.004388
4.92E+04 0.004467
5.08E+04 0.005132
5.24E+04 0.005721
5.57E+04 0.005736
5.90E+04 0.005191
6.23E+04 0.004775
6.55E+04 0.004999
6.88E+04 0.004409
7.21E+04 0.004058
7.54E+04 0.003633
7.86E+04 0.004695
8.19E+04 0.004972
8.52E+04 0.00504
8.85E+04 0.004858
9.18E+04 0.00484
9.50E+04 0.004306
9.83E+04 0.003788
1.02E+05 0.003981
1.05E+05 0.004506
1.11E+05 0.004171
1.18E+05 0.003168
1.25E+05 0.002522
1.31E+05 0.002944
1.38E+05 0.003594
1.44E+05 0.004418
1.51E+05 0.004553
1.57E+05 0.004326
1.64E+05 0.003709
1.70E+05 0.003877
1.77E+05 0.004466
1.84E+05 0.004872
1.90E+05 0.004488
1.97E+05 0.004045
2.03E+05 0.003579
2.10E+05 0.003666
2.23E+05 0.004628
2.36E+05 0.004984
2.49E+05 0.004944
2.62E+05 0.003804
2.75E+05 0.003782
2.88E+05 0.003964
3.02E+05 0.004853
3.15E+05 0.005339
3.28E+05 0.005248
3.41E+05 0.005212
3.54E+05 0.00545
3.67E+05 0.005813
3.80E+05 0.005659
3.93E+05 0.004797
4.06E+05 0.004675
4.19E+05 0.004457
4.46E+05 0.004952
4.72E+05 0.004876
4.98E+05 0.00543
5.24E+05 0.005862
5.51E+05 0.005677
5.77E+05 0.004955
6.03E+05 0.004261
6.29E+05 0.003847
6.55E+05 0.004229
6.82E+05 0.005012
7.08E+05 0.006114
7.34E+05 0.006305
7.60E+05 0.006245
7.86E+05 0.005616
8.13E+05 0.005305
8.39E+05 0.004986
8.91E+05 0.005342
9.44E+05 0.005732
9.96E+05 0.005743
1.05E+06 0.005812
1.10E+06 0.006329
1.15E+06 0.006729
1.21E+06 0.007516
1.26E+06 0.007742
1.31E+06 0.008152
1.36E+06 0.008627
1.42E+06 0.008485
1.47E+06 0.007884
1.52E+06 0.006802
1.57E+06 0.005282
1.63E+06 0.003002
1.68E+06 0.001459
1.78E+06 0
1.89E+06 0
1.99E+06 0
2.10E+06 0
2.20E+06 0
2.31E+06 0
2.41E+06 0
2.52E+06 0
2.62E+06 0
2.73E+06 0
2.83E+06 0
2.94E+06 0
3.04E+06 0
3.15E+06 0
3.25E+06 0
3.36E+06 0
3.57E+06 0
3.78E+06 0
3.99E+06 0
4.19E+06 0
4.40E+06 0
4.61E+06 0
4.82E+06 0
5.03E+06 0
5.24E+06 0
5.45E+06 0
5.66E+06 0
5.87E+06 0
6.08E+06 0
6.29E+06 0
6.50E+06 0
6.71E+06 0
7.13E+06 0
7.55E+06 0
7.97E+06 0
8.39E+06 0
8.81E+06 0
9.23E+06 0
9.65E+06 0
1.01E+07 0
1.05E+07 0
1.09E+07 0
1.13E+07 0
1.17E+07 0
1.22E+07 0
1.26E+07 0
1.30E+07 0
1.34E+07 0
1.43E+07 0
1.51E+07 0
1.59E+07 0
1.68E+07 0
1.76E+07 0
1.85E+07 0
1.93E+07 0
2.01E+07 0
2.10E+07 0
2.18E+07 0
2.27E+07 0
2.35E+07 0
2.43E+07 0
2.52E+07 0
2.60E+07 0
2.68E+07 0
2.85E+07 0
3.02E+07 0
3.19E+07 0
3.36E+07 0
3.52E+07 0
3.69E+07 0
3.86E+07 0
4.03E+07 0
4.19E+07 0
4.36E+07 0
4.53E+07 0
4.70E+07 0
4.87E+07 0
5.03E+07 0
5.20E+07 0
5.37E+07 0
5.70E+07 0
6.04E+07 0
6.38E+07 0
6.71E+07 0
7.05E+07 0
7.38E+07 0
7.72E+07 0
8.05E+07 0
8.39E+07 0
8.72E+07 0
9.06E+07 0
9.40E+07 0
9.73E+07 0
1.01E+08 0
1.04E+08 0
1.07E+08 0
1.14E+08 0
1.21E+08 0
1.28E+08 0
1.34E+08 0
1.41E+08 0
1.48E+08 0
1.54E+08 0
1.61E+08 0
1.68E+08 0
1.75E+08 0
1.81E+08 0
1.88E+08 0
1.95E+08 0
2.01E+08 0
2.08E+08 0
2.15E+08 0
"""


def parse_correlation_data(raw_data):
    time = []
    correlation = []

    for line in raw_data.strip().splitlines():
        parts = line.strip().split()

        if len(parts) < 2:
            continue

        try:
            time.append(float(parts[0]))
            correlation.append(float(parts[1]))
        except ValueError:
            continue

    return np.array(time), np.array(correlation)


def remove_zero_tail(time, correlation):
    """
    Removes the long zero tail at the end of the correlation curve.
    This keeps the real experimental decay and removes artificial zeros.
    """
    non_zero_indices = np.where(correlation != 0)[0]

    if len(non_zero_indices) == 0:
        return time, correlation

    last_non_zero = non_zero_indices[-1]

    return time[:last_non_zero + 1], correlation[:last_non_zero + 1]


time_run1, corr_run1 = parse_correlation_data(run1_raw)
time_run2, corr_run2 = parse_correlation_data(run2_raw)
time_run3, corr_run3 = parse_correlation_data(run3_raw)

time_run1, corr_run1 = remove_zero_tail(time_run1, corr_run1)
time_run2, corr_run2 = remove_zero_tail(time_run2, corr_run2)
time_run3, corr_run3 = remove_zero_tail(time_run3, corr_run3)

runs = [
    (time_run1, corr_run1, "Run 1", "#F2C94C"),
    (time_run2, corr_run2, "Run 2", "#F2994A"),
    (time_run3, corr_run3, "Run 3", "#EB5757")
]

plt.figure(figsize=(10, 5.8))

for time, corr, label, color in runs:
    plt.plot(
        time,
        corr,
        linewidth=2.3,
        color=color,
        label=label
    )

plt.xscale("log")

plt.xlabel("Time (µs)", fontsize=12)
plt.ylabel("Correlation coefficient ($g_2 - 1$)", fontsize=12)
plt.title("Sample 1: Correlation function", fontsize=14, fontweight="bold")

plt.grid(True, which="major", linestyle="--", alpha=0.5)
plt.grid(True, which="minor", linestyle=":", alpha=0.25)

plt.legend(frameon=True)

plt.tight_layout()
plt.show()
```
